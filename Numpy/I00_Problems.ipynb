#### 28. What are the result of the following expressions? (★☆☆)
```python
np.array(0) / np.array(0)
np.array(0) // np.array(0)
np.array([np.nan]).astype(int).astype(float)
```

````python
# 0/0 is mathematically undefined.
# NumPy returns NaN (Not a Number) instead of crashing.
# NaN means "this computation has no valid numeric result."
np.array(0) / np.array(0)

# '//' performs floor division(goes to the nearest whole number after / and that's the solution)
# But 0//0 is still an undefined operation.
# NumPy again returns NaN because the result doesn't exist.
np.array(0) // np.array(0)

# NaN represents a missing/undefined floating-point value.
# Integers cannot store NaN.
# Converting NaN to int loses the missing-value information
# (or raises an error in modern NumPy versions).
# Converting back to float DOES NOT restore NaN.
np.array([np.nan]).astype(int).astype(float)

# NaN = "I don't know this value."
# It is NOT zero, NOT empty, and NOT None.
# It represents a missing or undefined numeric value.

# astype() changes only the data type.
# If the new data type cannot represent the value,
# information may be lost or an error may occur.

# Missing values (NaN) are extremely common in real datasets.
# Always handle NaN before training ML models or converting data types.
````

#### 50. How to find the closest value (to a given scalar) in a vector? (★★☆)

````python
#Array
#  ↓
#Find distance from scalar by diffrence (arr-n)
#            ↓
#Take absolute value (do -ve's into +ve)
#        ↓
#Find smallest distance (argmin)
#              ↓
#Use that index on the original array

arr=np.array((12,11,17,21,23,28))
print(arr)

scalar=15  #this could be any number and if the same in the array then it will give that number

Diff=arr-scalar  #by this we will get the closest shi to our value
print(Diff)

getabs=np.abs(Diff)  #this 'abs' gives us absolute value without caring about signs(+,-)
print(getabs)  #it gives all the -ve values into +ve cuz all we care about is distance

findClosest=np.argmin(getabs)  #now to get the index value of the closest index i.e smallest idx
print(findClosest)

el=arr[findClosest]  #finally in the array i will get the value by it's index that i've found
print(el)

print("The Closest El in the array to our Scalar",scalar,"is",el)
````

#### 75. How to compute averages using a sliding window over an array? (★★★)

````python
#Sliding window is a point till which we have to perform the action(here they said for avges),
#so i gotta cal only of first 'SW input' el's averages and then i gotta take a 1 step next and 
#then cal. the next three until the last tip touches the end of the arrray.
#So, ar=[123456], Sw=2 then we go 12,23,34,45,56 and of these pairs we return the avg of each.

arr=np.array((3,4,6,8,2,7))
SW=3

kernel=np.ones(SW)/SW  #will give -> 1/3,1/3,1/3 (creating space which will get added later)

#This slides the kernel^ over an arr. At each group it *el w 1 ->sum the products->stores result->moves kernel +1
final=np.convolve(arr, kernel, mode='valid')  #valid mode runs till the end'th idx
print(final)  #kernel stores space of SWinput(3) & asks convolve to apply first 3el then iterates till array's end

# 1.kerleno defines the operation to perform on how manny of em 
# 2.Convolve slides the kernel over the array while performing mul->sum->stores and repeats for every itr till end



from numpy.lib.stride_tricks import sliding_window_view  #Another way(literal)

arr1=np.array((3,4,6,8,6,7))  
sw=3  

window=sliding_window_view(arr1, window_shape=sw)  #everything is given and this divides into that much sw
print(window) #i got all the possible windows till end, now what's left is to find the fukking mean of all these 4

resultingGroup=window.mean(axis=1) #agian axis=0 is row wise and a=1 is columnwize(i.e 1st el of each row) and so
print(resultingGroup)  #i got the final result but it is such a hassle to memorize this much..
````

#### 91. How to create a record array from a regular array? (★★★)

````python
#Question wants to create somthing like a dictionary like -:
#print(arr[:,1],arr[:,0]) #see the array below , this is what they want but after giving name to it
#normally i have to acces with using indexing(:1) but i need to make somthing so the each
#column has a field name so i can access by name instead of by column idx 

arr = np.array([    #here we got 3 Records & 2 Data Column of each of those records
    [1, 90],
    [2, 85],     #Step 1 Done - Creating an Array
    [3, 95]
])

# i want to transpose the array first cuz array shape is 3*2 that means each column have 2 data,
# but i want to access like only the first part or 2nd part i.e our id and marks so we gotta need
# to trasform into 2*3 i.e all the ides on 1 array and all the marks on 2nd array. Also 'fromarray' 
# (this is the func. that we are going to use next) excepts 1 array at a time so it can link 1st array
# columns to 2nd array's columns.
transposed = arr.T   #So the shape is 2*3 now

# Now we need to name those 2 data arrays also we gotta specify their type. One is Id and the other is marks
# also deep analogy of the datatype is that what if a column contatins diff dtype values so that's why it's 
# like this. Also these names lets us call those specific parts.
CusDtype =[ ("id",int) , ("marks",int) ] #making a custom dtype which have 2 parameter for diff columns/arrays


#fromarrays divides and connects array as we know so we can access it from 'id' & 'marks'
recordd = np.core.records.fromarrays(transposed, dtype=CusDtype) #passed those 2 parameters that we build
print(recordd)  #see how it will look like and now i just gotta use '.name' to access perticular column

print(recordd.id)  #from here i can access 1 column whichever i want
print(recordd.marks)   #this 'core.records' came from very deep so we need 'deep'
#So,we created a record array from these arrays, and use this dtype to name and defined the columns.
````

#### 97. Considering 2 vectors A & B, write the einsum equivalent of inner, outer, sum, and mul function (★★★)

````python
# Question is saying without using normal 'np.inner'(cal the mul of center el of both array),
# 'np.outer'(ye pta ni kya hai), 'np.sum' and "np.mul" etc functions. So i don't have to 
# directly use these. I have to use this 'einsum' somehow and get the result of all of em

vec1=np.random.randint(1,8,5)   #made 2 1D vectors for the operations
vec2=np.random.randint(5,12,5) 
print(vec1,vec2)                #einsum stands from einstien sum btw


#Prereqiseites:
# using 'i,i->' means 'inner' cuz inner do arr1*arr2 and then add + between all of those so it returns only 1 value.
# using 'i,j->ij' means 'outer' cuz it compares arr's 1st el w each el of 2nd ar->2nd el of 1st ar w each el of 2nd ar..
# using 'i->' means just sum cuz it have 1 i that means only 1 vec will be passed in 2nd parameter while esuming
# using 'i,i->i' means 'arr1*arr2' cuz each arr have it's 'i' and they are written together and they return a i.

inner=np.einsum('i,i->',vec1,vec2)  #when keeping nothing after '->' that always means we'll get 1 final value i.e sum
print("Inner sum of Vec1 & Vec2 is",inner)  

outer=np.einsum('i,j->ij',vec1,vec2)  #here i need manny rows and some combinations so diff varibles and returning ij
print(outer) #it means making 3*3 combination out of two 1D arrays both having size of 3

summ1=np.einsum('i->',vec1)  #finding sum of each vector individually and as i said if nothing after -> that means sum
sum2=np.einsum('j->',vec2)  #i forgor to mention that i could use anyvarible  but i prefer same when possible
print("Total in 1st Vector :",summ1,"In 2nd it's :",sum2)

mul=np.einsum('i,i->i',vec1,vec2)  #this gets me a array in return that is the multiplication of each el in both array
print(mul)  # like here i have to do i,i not choosing j of my own cuz it'g giving wrong result with j so prefer i shi
````
