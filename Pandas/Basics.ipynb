26. Write a Function that can return the track record of 2 team against each other

```python
def fight(first, second):
    t1= ipl['Team1'] == first
    t2= ipl['Team2'] == second   #Both possible team orders are considered.
    t3= ipl['Team1'] == second
    t4= ipl['Team2'] == first

    return ipl[(t1 & t2) | (t3 & t4)]  #Select matches where two teams played against each other in any order

fight('Rajasthan Royals', 'Kolkata Knight Riders')
```
